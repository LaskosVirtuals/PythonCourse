# Reading Class Definitions

AI generates classes constantly. Let's decode every part of a class definition.

## The Anatomy of a Class

```python
class ClassName(ParentClass):      # Class definition with inheritance
    """Docstring: what this class does."""
    
    class_attribute = "shared"      # Shared by ALL instances
    
    def __init__(self, arg):        # Constructor
        self.instance_attr = arg    # Unique to each instance
    
    def method(self):               # Instance method
        return self.instance_attr
    
    @classmethod                    # Class method
    def from_string(cls, s):
        return cls(s)
    
    @staticmethod                   # Static method
    def utility():
        return "no self needed"
```

````python
# Let's build a class piece by piece

class Dog:
    """A simple Dog class to demonstrate OOP concepts."""
    
    # Class attribute - shared by ALL dogs
    species = "Canis familiaris"
    
    def __init__(self, name, age):
        """Constructor - runs when you create a new Dog."""
        # Instance attributes - unique to each dog
        self.name = name
        self.age = age
    
    def bark(self):
        """Instance method - uses self."""
        return f"{self.name} says woof!"
    
    def __str__(self):
        """Called when you print() or str() the object."""
        return f"Dog({self.name}, {self.age} years old)"

# Create instances
buddy = Dog("Buddy", 3)
max_dog = Dog("Max", 5)

print(buddy.bark())
print(max_dog.bark())
print(f"Species: {buddy.species}")  # Class attribute
print(buddy)  # Uses __str__
````

## Understanding `self`

**`self` is the instance** - it's how methods access the object's data.

````python
class Counter:
    def __init__(self):
        self.count = 0  # 'self' refers to THIS specific counter
    
    def increment(self):
        self.count += 1  # Access instance data via self
        return self.count

# Two separate counters, each with its own 'count'
counter1 = Counter()
counter2 = Counter()

print(f"Counter 1: {counter1.increment()}")  # 1
print(f"Counter 1: {counter1.increment()}")  # 2
print(f"Counter 2: {counter2.increment()}")  # 1 (separate!)
````

## Class vs Instance Attributes

````python
class Example:
    class_attr = "I'm shared"  # Class attribute
    
    def __init__(self, value):
        self.instance_attr = value  # Instance attribute

obj1 = Example("one")
obj2 = Example("two")

# Instance attributes are separate
print(f"obj1.instance_attr: {obj1.instance_attr}")  # "one"
print(f"obj2.instance_attr: {obj2.instance_attr}")  # "two"

# Class attribute is shared
print(f"obj1.class_attr: {obj1.class_attr}")  # "I'm shared"
print(f"obj2.class_attr: {obj2.class_attr}")  # "I'm shared"

# Changing class attribute affects all!
Example.class_attr = "Changed!"
print(f"obj1.class_attr: {obj1.class_attr}")  # "Changed!"
print(f"obj2.class_attr: {obj2.class_attr}")  # "Changed!"
````

## Method Types: Regular, Class, Static

````python
class MyClass:
    class_data = "shared"
    
    def __init__(self, value):
        self.value = value
    
    # Regular instance method - has 'self'
    def instance_method(self):
        return f"Instance data: {self.value}"
    
    # Class method - has 'cls' instead of 'self'
    @classmethod
    def class_method(cls):
        return f"Class data: {cls.class_data}"
    
    # Alternative constructor pattern
    @classmethod
    def from_string(cls, string):
        """Create instance from a string."""
        return cls(string.upper())
    
    # Static method - no self or cls
    @staticmethod
    def static_method(x, y):
        return x + y

obj = MyClass("hello")

print(obj.instance_method())      # Uses self.value
print(MyClass.class_method())     # Uses cls.class_data
print(MyClass.static_method(1, 2)) # Pure function

# Alternative constructor
obj2 = MyClass.from_string("test")
print(obj2.value)  # "TEST"
````

## Private Attributes (Convention)

Python doesn't have true private attributes, but uses naming conventions:

````python
class BankAccount:
    def __init__(self, balance):
        self._balance = balance      # Single underscore: "private by convention"
        self.__secret = "hidden"     # Double underscore: name mangling
    
    def get_balance(self):
        return self._balance

account = BankAccount(1000)

# Single underscore - accessible but "please don't"
print(account._balance)  # Works, but you shouldn't

# Double underscore - name mangled
# print(account.__secret)  # AttributeError!
print(account._BankAccount__secret)  # Works but really don't do this
````

## Type Hints in Classes

Modern Python and AI code uses type hints extensively:

````python
from typing import Optional, List

class User:
    def __init__(self, name: str, age: int, email: Optional[str] = None):
        self.name: str = name
        self.age: int = age
        self.email: Optional[str] = email
        self.friends: List[str] = []
    
    def add_friend(self, friend_name: str) -> None:
        self.friends.append(friend_name)
    
    def get_info(self) -> str:
        return f"{self.name}, {self.age} years old"

user = User("Alice", 30, "alice@example.com")
user.add_friend("Bob")
print(user.get_info())
print(f"Friends: {user.friends}")
````

## Summary

| Concept | Syntax | Purpose |
|---------|--------|--------|
| Class definition | `class Name:` | Create a new type |
| Constructor | `def __init__(self):` | Initialize instance |
| Instance attribute | `self.attr = value` | Data unique to instance |
| Class attribute | `attr = value` (in class body) | Data shared by all |
| Instance method | `def method(self):` | Works with instance data |
| Class method | `@classmethod` + `cls` | Works with class data |
| Static method | `@staticmethod` | No instance/class access |
| Private convention | `_attr` or `__attr` | "Don't touch" signal |
