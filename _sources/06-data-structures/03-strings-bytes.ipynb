# Strings and Bytes

Understanding the difference between text (str) and binary data (bytes) is crucial for working with files and networks.

## String Basics

```python
# Creating strings
single = 'Hello'
double = "World"
triple = '''Multi
line
string'''
raw = r"C:\Users\name"  # Raw string (no escapes)

print(f"Single: {single}")
print(f"Double: {double}")
print(f"Triple: {triple}")
print(f"Raw: {raw}")
```

```python
# F-strings (formatted string literals)
name = "Alice"
age = 30

# Basic interpolation
print(f"Name: {name}, Age: {age}")

# Expressions
print(f"Next year: {age + 1}")

# Formatting
pi = 3.14159
print(f"Pi: {pi:.2f}")          # 2 decimal places
print(f"Padded: {42:05d}")      # Zero-padded
print(f"Aligned: {name:>10}")   # Right-aligned
print(f"Hex: {255:#x}")         # Hexadecimal
```

```python
# String methods
text = "  Hello, World!  "

print(f"strip: '{text.strip()}'")
print(f"lower: '{text.lower()}'")
print(f"upper: '{text.upper()}'")
print(f"replace: '{text.replace('World', 'Python')}'")
print(f"split: {text.split()}")
print(f"startswith: {text.strip().startswith('Hello')}")
```

```python
# Joining strings
words = ["apple", "banana", "cherry"]

print(f"Join with comma: {', '.join(words)}")
print(f"Join with newline:")
print('\n'.join(words))
```

## Bytes vs Strings

```python
# str = text (Unicode characters)
# bytes = binary data (raw bytes)

text = "Hello, World!"  # str
binary = b"Hello, World!"  # bytes (ASCII only)

print(f"String type: {type(text)}")
print(f"Bytes type: {type(binary)}")

print(f"String: {text}")
print(f"Bytes: {binary}")
```

```python
# Converting between str and bytes
text = "Hello, 世界!"  # Unicode text

# Encode: str -> bytes
encoded = text.encode('utf-8')
print(f"Encoded: {encoded}")
print(f"Bytes: {list(encoded)}")

# Decode: bytes -> str
decoded = encoded.decode('utf-8')
print(f"Decoded: {decoded}")
```

```python
# Different encodings
text = "café"

utf8 = text.encode('utf-8')
latin1 = text.encode('latin-1')

print(f"UTF-8: {utf8} (length: {len(utf8)})")
print(f"Latin-1: {latin1} (length: {len(latin1)})")

# UTF-8 uses more bytes for non-ASCII
```

```python
# Handling encoding errors
text = "Hello, 世界!"

# This fails: ASCII can't encode Chinese
try:
    text.encode('ascii')
except UnicodeEncodeError as e:
    print(f"Error: {e}")

# Options for handling:
print(f"ignore: {text.encode('ascii', errors='ignore')}")
print(f"replace: {text.encode('ascii', errors='replace')}")
print(f"xmlcharrefreplace: {text.encode('ascii', errors='xmlcharrefreplace')}")
```

## When to Use What?

| Use `str` | Use `bytes` |
|-----------|-------------|
| Text files | Binary files |
| JSON data | Images, audio |
| User input | Network sockets |
| Database text | Cryptography |
| Display | Raw file I/O |

## String Methods Deep Dive

```python
# Searching
text = "Hello, World! Hello, Python!"

print(f"find 'World': {text.find('World')}")
print(f"find 'Java': {text.find('Java')}")  # -1 if not found
print(f"count 'Hello': {text.count('Hello')}")
print(f"'Hello' in text: {'Hello' in text}")
```

```python
# Checking string content
print(f"'hello'.isalpha(): {'hello'.isalpha()}")
print(f"'123'.isdigit(): {'123'.isdigit()}")
print(f"'abc123'.isalnum(): {'abc123'.isalnum()}")
print(f"'   '.isspace(): {'   '.isspace()}")
print(f"'Hello'.istitle(): {'Hello World'.istitle()}")
```

```python
# Splitting and parsing
csv_line = "apple,banana,cherry"
path = "/home/user/documents/file.txt"

print(f"Split CSV: {csv_line.split(',')}")
print(f"Split path: {path.split('/')}")

# splitlines for multiline text
multiline = "line1\nline2\nline3"
print(f"Lines: {multiline.splitlines()}")

# partition - split into 3 parts
text = "key=value"
print(f"Partition: {text.partition('=')}")
```

## Regular Expressions (Brief Intro)

```python
import re

text = "Contact: alice@example.com or bob@test.org"

# Find all email addresses
pattern = r'[\w.+-]+@[\w-]+\.[\w.-]+'
emails = re.findall(pattern, text)
print(f"Found emails: {emails}")

# Search for pattern
match = re.search(r'\d+', "Order 12345")
if match:
    print(f"Found number: {match.group()}")

# Replace
cleaned = re.sub(r'\s+', ' ', "Too    many   spaces")
print(f"Cleaned: {cleaned}")
```

```python
# Pattern 2: Template strings
from string import Template

template = Template("Hello, $name! You have $count messages.")
result = template.substitute(name="Alice", count=5)
print(result)

# Safe substitute (missing vars become empty)
result = template.safe_substitute(name="Bob")
print(result)
```

```python
# Pattern 3: Building strings efficiently
# BAD: String concatenation in loop (slow)
# result = ""
# for item in items:
#     result += str(item)  # Creates new string each time!

# GOOD: Use join
items = range(10)
result = ''.join(str(item) for item in items)
print(f"Joined: {result}")

# GOOD: Use list and join
parts = []
for i in range(5):
    parts.append(f"item{i}")
result = ', '.join(parts)
print(f"Parts: {result}")
```

```python
# Pattern 4: Normalize strings for comparison
def normalize(s):
    """Normalize string for comparison."""
    return s.lower().strip()

user_input = "  HELLO  "
expected = "hello"

print(f"Match: {normalize(user_input) == expected}")
```

## Bytes Operations

```python
# bytearray - mutable bytes
data = bytearray(b"Hello")
data[0] = ord('J')  # Replace first byte
print(f"Modified: {data}")

# bytes are immutable
frozen = bytes(data)
# frozen[0] = ord('H')  # TypeError!
```

```python
# Hex encoding
data = b"Hello"

hex_string = data.hex()
print(f"Hex: {hex_string}")

back = bytes.fromhex(hex_string)
print(f"From hex: {back}")
```

```python
# Base64 encoding (common in APIs)
import base64

data = b"Hello, World!"

encoded = base64.b64encode(data)
print(f"Base64: {encoded}")

decoded = base64.b64decode(encoded)
print(f"Decoded: {decoded}")
```
