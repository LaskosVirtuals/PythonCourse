# Core Collections

Python's four fundamental collection types: lists, dicts, sets, and tuples.

## Lists - Ordered, Mutable Sequences

```python
# Creating lists
empty = []
numbers = [1, 2, 3, 4, 5]
mixed = [1, "hello", 3.14, True, None]  # Any types
nested = [[1, 2], [3, 4], [5, 6]]

print(f"Numbers: {numbers}")
print(f"Mixed: {mixed}")
print(f"Nested: {nested}")
```

```python
# Indexing and slicing
items = ["a", "b", "c", "d", "e"]

print(f"First: {items[0]}")
print(f"Last: {items[-1]}")
print(f"Slice [1:3]: {items[1:3]}")
print(f"Every 2nd: {items[::2]}")
print(f"Reversed: {items[::-1]}")
```

```python
# Modifying lists
items = [1, 2, 3]

items.append(4)          # Add to end
print(f"After append: {items}")

items.insert(0, 0)       # Insert at index
print(f"After insert: {items}")

items.extend([5, 6])     # Add multiple
print(f"After extend: {items}")

removed = items.pop()    # Remove & return last
print(f"Popped: {removed}, List: {items}")

items.remove(0)          # Remove first occurrence
print(f"After remove: {items}")
```

```python
# List operations
a = [1, 2, 3]
b = [4, 5, 6]

print(f"Concatenate: {a + b}")
print(f"Repeat: {a * 3}")
print(f"Length: {len(a)}")
print(f"3 in a: {3 in a}")
print(f"Sum: {sum(a)}")
print(f"Min/Max: {min(a)}/{max(a)}")
```

## Dictionaries - Key-Value Mappings

```python
# Creating dicts
empty = {}
user = {"name": "Alice", "age": 30, "active": True}
from_pairs = dict([("a", 1), ("b", 2)])
from_kwargs = dict(x=1, y=2, z=3)

print(f"User: {user}")
print(f"From pairs: {from_pairs}")
print(f"From kwargs: {from_kwargs}")
```

```python
# Accessing values
user = {"name": "Alice", "age": 30}

print(f"Name: {user['name']}")
# print(user['email'])  # KeyError!

# Safe access with .get()
print(f"Email: {user.get('email')}")
print(f"Email: {user.get('email', 'not set')}")
```

```python
# Modifying dicts
user = {"name": "Alice"}

user["age"] = 30               # Add/update key
user.update({"role": "admin"}) # Merge dict
print(f"Updated: {user}")

del user["role"]              # Delete key
age = user.pop("age")         # Remove & return
print(f"After removal: {user}")
```

```python
# Iterating dicts
config = {"host": "localhost", "port": 8080, "debug": True}

print("Keys:")
for key in config:
    print(f"  {key}")

print("\nValues:")
for value in config.values():
    print(f"  {value}")

print("\nItems:")
for key, value in config.items():
    print(f"  {key}: {value}")
```

```python
# setdefault - get or set if missing
counts = {}

# Without setdefault
# if "a" not in counts:
#     counts["a"] = 0
# counts["a"] += 1

# With setdefault
counts.setdefault("a", 0)
counts["a"] += 1

print(f"Counts: {counts}")
```

## Sets - Unique Unordered Collections

```python
# Creating sets
empty = set()  # NOT {} - that's an empty dict!
numbers = {1, 2, 3, 4, 5}
from_list = set([1, 2, 2, 3, 3, 3])  # Duplicates removed

print(f"Numbers: {numbers}")
print(f"From list: {from_list}")
```

```python
# Set operations
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}

print(f"Union (a | b): {a | b}")
print(f"Intersection (a & b): {a & b}")
print(f"Difference (a - b): {a - b}")
print(f"Symmetric diff (a ^ b): {a ^ b}")
```

```python
# Set membership (very fast!)
valid_statuses = {"pending", "active", "completed", "cancelled"}

status = "active"
print(f"'{status}' is valid: {status in valid_statuses}")

status = "invalid"
print(f"'{status}' is valid: {status in valid_statuses}")
```

```python
# Modifying sets
items = {1, 2, 3}

items.add(4)          # Add one
items.update([5, 6])  # Add multiple
print(f"After adding: {items}")

items.discard(6)      # Remove (no error if missing)
items.remove(5)       # Remove (error if missing)
print(f"After removing: {items}")
```

## Tuples - Immutable Sequences

```python
# Creating tuples
empty = ()
single = (1,)  # Note the comma!
point = (3, 4)
mixed = (1, "hello", 3.14)

print(f"Single: {single}")
print(f"Point: {point}")
print(f"Not a tuple: {type((1))}")
print(f"Is a tuple: {type((1,))}")
```

```python
# Tuples are immutable
point = (3, 4)
# point[0] = 5  # TypeError!

# But can contain mutable objects
data = ([1, 2], [3, 4])
data[0].append(3)  # This works!
print(f"Data: {data}")
```

```python
# Tuple unpacking
point = (3, 4)
x, y = point
print(f"x={x}, y={y}")

# With * for remaining
first, *rest = (1, 2, 3, 4, 5)
print(f"First: {first}, Rest: {rest}")

*beginning, last = (1, 2, 3, 4, 5)
print(f"Beginning: {beginning}, Last: {last}")
```

```python
# Tuples as dict keys (lists can't be!)
locations = {
    (0, 0): "origin",
    (1, 0): "east",
    (0, 1): "north",
}

print(f"At (0,0): {locations[(0, 0)]}")

# Can't do this with lists:
# {[0, 0]: "origin"}  # TypeError: unhashable type: 'list'
```

## Named Tuples

```python
from collections import namedtuple

# Define a named tuple type
Point = namedtuple("Point", ["x", "y"])
User = namedtuple("User", "name age email")  # String syntax also works

# Create instances
p = Point(3, 4)
u = User("Alice", 30, "alice@example.com")

# Access by name or index
print(f"Point: x={p.x}, y={p.y}")
print(f"User: {u.name} ({u.age})")
print(f"By index: {p[0]}, {p[1]}")
```

## When to Use What?

| Type | Use When |
|------|----------|
| `list` | Ordered items, may change, duplicates OK |
| `dict` | Key-value pairs, fast lookup by key |
| `set` | Unique items, fast membership testing |
| `tuple` | Fixed data, dict keys, function returns |

```python
# Common pattern: Remove duplicates while preserving order
items = [3, 1, 4, 1, 5, 9, 2, 6, 5, 3, 5]

# Using set (loses order)
unique_unordered = list(set(items))
print(f"Set (unordered): {unique_unordered}")

# Using dict (preserves order in Python 3.7+)
unique_ordered = list(dict.fromkeys(items))
print(f"Dict (ordered): {unique_ordered}")
```

## Summary

| Type | Syntax | Mutable | Ordered | Duplicates |
|------|--------|---------|---------|------------|
| List | `[1, 2, 3]` | Yes | Yes | Yes |
| Dict | `{"a": 1}` | Yes | Yes* | Keys: No |
| Set | `{1, 2, 3}` | Yes | No | No |
| Tuple | `(1, 2, 3)` | No | Yes | Yes |

*Dicts preserve insertion order since Python 3.7
