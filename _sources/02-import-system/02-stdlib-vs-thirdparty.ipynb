# Standard Library vs. Third-Party Packages

When AI generates code with imports, a common question is:

> "Do I need to install something, or is this built into Python?"

This lesson teaches you to tell the difference.

## The Three Sources of Imports

Python code can import from three places:

| Source | Installation | Example |
|--------|-------------|--------|
| **Standard Library** | Built-in, nothing to install | `import os`, `import json` |
| **Third-Party** | Install with `pip install` | `import requests`, `import pandas` |
| **Local/Project** | Your own code | `from myapp import utils` |

## The Standard Library

Python comes with **batteries included** - a huge collection of useful modules. These are always available.

### Common Standard Library Modules You'll See

````python
# FILE AND PATH OPERATIONS
import os           # Operating system interface
import pathlib      # Modern path handling (Path objects)
import shutil       # High-level file operations (copy, move)
import glob         # Find files matching patterns

# These are ALWAYS available - no pip install needed
print(f"Current directory: {os.getcwd()}")
print(f"Using pathlib: {pathlib.Path.cwd()}")
````

````python
# DATA FORMATS
import json         # JSON encoding/decoding
import csv          # CSV file handling
import xml          # XML processing
import sqlite3      # SQLite database

# JSON is everywhere in AI code
data = {"name": "Claude", "type": "AI"}
json_string = json.dumps(data, indent=2)
print(json_string)
````

````python
# DATE AND TIME
import datetime     # Date and time types
import time         # Time-related functions
import calendar     # Calendar operations

now = datetime.datetime.now()
print(f"Current time: {now}")
print(f"Formatted: {now.strftime('%Y-%m-%d %H:%M:%S')}")
````

````python
# TEXT AND PATTERNS
import re           # Regular expressions
import string       # String constants
import textwrap     # Text wrapping and formatting

# Regular expressions - AI loves these
text = "Contact: email@example.com or call 123-456-7890"
emails = re.findall(r'\S+@\S+', text)
print(f"Found emails: {emails}")
````

````python
# DATA STRUCTURES
from collections import Counter, defaultdict, deque
import itertools    # Iteration tools
import functools    # Higher-order functions

# Counter is super useful
words = ["apple", "banana", "apple", "cherry", "banana", "apple"]
counts = Counter(words)
print(f"Word counts: {counts}")
print(f"Most common: {counts.most_common(2)}")
````

````python
# SYSTEM AND RUNTIME
import sys          # System-specific parameters
import subprocess   # Run external commands
import argparse     # Command-line argument parsing
import logging      # Logging facility

print(f"Python version: {sys.version}")
print(f"Python executable: {sys.executable}")
````

````python
# TYPE HINTS (very common in AI code)
from typing import Optional, List, Dict, Union, Callable

# These don't do anything at runtime - they're for documentation
def greet(name: str, times: int = 1) -> List[str]:
    """Return a list of greetings."""
    return [f"Hello, {name}!"] * times

result = greet("World", 3)
print(result)
````

## Third-Party Packages

These are NOT built into Python. They must be installed.

### How to Know If Something Is Third-Party

**Quick check:** If it's not in the [Python Standard Library docs](https://docs.python.org/3/library/), it's third-party.

### Common Third-Party Packages AI Uses

```python

```

### The `ModuleNotFoundError`

When you try to import something that isn't installed:

````python
# This will likely fail in the browser (Pyodide doesn't have requests)
# But you'll see this error often in local Python

try:
    import some_package_that_doesnt_exist
except ModuleNotFoundError as e:
    print(f"Error: {e}")
    print("")
    print("This error means: the package isn't installed.")
    print("Fix it with: pip install package_name")
````

## How AI Handles Missing Packages

AI often generates **defensive imports** - code that handles missing packages gracefully:

````python
# Pattern 1: Try/except with helpful message
try:
    import requests
except ImportError:
    raise ImportError(
        "The 'requests' package is required. "
        "Install it with: pip install requests"
    )
````

````python
# Pattern 2: Optional import with fallback
try:
    import ujson as json  # Faster JSON library
except ImportError:
    import json  # Fall back to standard library

print(f"Using: {json.__name__}")
````

````python
# Pattern 3: Feature flag based on availability
try:
    import numpy as np
    HAS_NUMPY = True
except ImportError:
    HAS_NUMPY = False

def process_data(data):
    if HAS_NUMPY:
        return np.array(data).mean()
    else:
        return sum(data) / len(data)

print(f"NumPy available: {HAS_NUMPY}")
print(f"Average of [1,2,3,4,5]: {process_data([1,2,3,4,5])}")
````

## Quick Reference: Is It Standard Library?

### Almost Always Standard Library (No Install)

| Module | Purpose |
|--------|--------|
| `os`, `sys`, `pathlib` | System/files |
| `json`, `csv`, `xml` | Data formats |
| `datetime`, `time` | Dates/times |
| `re` | Regular expressions |
| `collections`, `itertools` | Data structures |
| `typing` | Type hints |
| `logging` | Logging |
| `unittest`, `doctest` | Testing |
| `http`, `urllib` | Basic HTTP |
| `sqlite3` | SQLite database |
| `math`, `random`, `statistics` | Math |
| `subprocess`, `threading`, `multiprocessing` | Concurrency |

### Almost Always Third-Party (Need Install)

| Module | Install Command |
|--------|----------------|
| `requests` | `pip install requests` |
| `numpy` / `np` | `pip install numpy` |
| `pandas` / `pd` | `pip install pandas` |
| `flask`, `fastapi`, `django` | `pip install flask` etc. |
| `pydantic` | `pip install pydantic` |
| `pytest` | `pip install pytest` |
| `beautifulsoup4` / `bs4` | `pip install beautifulsoup4` |

````python
# Check if a module is standard library or third-party
import json
import sys

# Standard library modules are usually in the Python installation directory
print(f"json module location: {json.__file__}")

# If __file__ contains 'site-packages', it's third-party
# If it's in the Python lib directory, it's standard library
````

## Summary

When you see an import in AI-generated code:

1. **Check if it's standard library** - look at the reference table above
2. **If not sure** - Google "python [module name]" or check the [Python docs](https://docs.python.org/3/library/)
3. **If you get `ModuleNotFoundError`** - run `pip install module_name`
4. **Look for defensive imports** - AI often includes try/except for optional packages
