# Project Structure Patterns

When AI generates a project, it follows common patterns. Understanding these patterns helps you navigate any Python codebase.

## The Three Common Layouts

### Layout 1: Flat (Simple Scripts)

For small projects or single scripts:

```
my_project/
    main.py           # Entry point
    utils.py          # Helper functions
    config.py         # Configuration
    requirements.txt  # Dependencies
```

**When AI uses this:** Quick scripts, prototypes, simple automation tasks.

**How to run:** `python main.py`

### Layout 2: Package in Root

For medium projects with a clear package:

```
my_project/
    mypackage/
        __init__.py   # Makes it a package
        core.py       # Main logic
        utils.py      # Utilities
    tests/
        test_core.py
    setup.py          # Installation script
    requirements.txt
```

**When AI uses this:** Libraries, tools meant to be imported.

**How to run:** `python -m mypackage` or import it

### Layout 3: src Layout (Modern Best Practice)

For professional projects:

```
my_project/
    src/
        mypackage/
            __init__.py
            core.py
            cli.py
    tests/
        test_core.py
    pyproject.toml    # Modern config
    README.md
```

**Why `src/`?** Prevents accidentally importing the wrong code during development.

**When AI uses this:** Serious projects, packages for PyPI.

## Key Files Explained

| File | Purpose |
|------|--------|
| `__init__.py` | Marks folder as a Python package |
| `__main__.py` | Allows `python -m package` execution |
| `requirements.txt` | Lists dependencies for `pip install -r` |
| `pyproject.toml` | Modern project configuration |
| `setup.py` | Legacy installation script |
| `.gitignore` | Files Git should ignore |
| `README.md` | Project documentation |

## Common Folders

| Folder | Contents |
|--------|----------|
| `src/` | Source code (in src layout) |
| `tests/` | Test files |
| `docs/` | Documentation |
| `scripts/` | Utility scripts |
| `data/` | Data files |
| `config/` | Configuration files |
| `.github/` | GitHub Actions workflows |

````python
# Let's explore a real project structure
import os

def show_structure(path=".", prefix=""):
    """Display folder structure like the 'tree' command."""
    items = sorted(os.listdir(path))
    # Filter out hidden and common ignore patterns
    items = [i for i in items if not i.startswith('.') 
             and i not in ['__pycache__', 'venv', 'node_modules']]
    
    for i, item in enumerate(items):
        is_last = (i == len(items) - 1)
        connector = "└── " if is_last else "├── "
        print(f"{prefix}{connector}{item}")
        
        item_path = os.path.join(path, item)
        if os.path.isdir(item_path):
            extension = "    " if is_last else "│   "
            show_structure(item_path, prefix + extension)

# This would show your current folder structure
# show_structure(".")
print("Function defined! Use show_structure('.') to explore a folder.")
````

## AI Coder Tip: Recognizing Project Types

When AI generates code, look for these clues:

**Web Application:**
```
app.py or main.py with Flask/FastAPI
templates/
static/
```

**CLI Tool:**
```
cli.py or __main__.py
[project.scripts] in pyproject.toml
```

**Data Science:**
```
notebooks/
data/
models/
```

**Library/Package:**
```
src/ layout
__init__.py with exports
pyproject.toml with [build-system]
```

## Summary

| Layout | Best For | Key Indicator |
|--------|----------|---------------|
| Flat | Scripts, prototypes | No package folder |
| Package in root | Libraries | `mypackage/` at root |
| src layout | Professional projects | `src/mypackage/` |
