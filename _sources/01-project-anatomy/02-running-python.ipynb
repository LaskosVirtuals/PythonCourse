# Running Python Code

AI generates code, but how do you actually run it? This lesson covers all the ways to execute Python.

## The Different Ways to Run Python

| Command | What It Does |
|---------|-------------|
| `python script.py` | Run a script file |
| `python -m module` | Run a module as a script |
| `python -c "code"` | Run code directly |
| `python` | Start interactive interpreter |

## Understanding `if __name__ == "__main__"`

You'll see this at the bottom of almost every AI-generated script:

```python
def main():
    print("Hello, World!")

if __name__ == "__main__":
    main()
```

**What does this mean?**

````python
# Every Python file has a special variable: __name__
print(f"This file's __name__ is: {__name__}")

# When you RUN a file directly: __name__ == "__main__"
# When you IMPORT a file: __name__ == the module name
````

````python
# Example: Why this pattern matters

# Imagine this is utils.py:
def calculate_sum(numbers):
    return sum(numbers)

def test_calculate_sum():
    result = calculate_sum([1, 2, 3])
    print(f"Test result: {result}")
    assert result == 6, "Test failed!"
    print("Test passed!")

# This code ONLY runs if you execute utils.py directly
# It does NOT run if someone imports utils.py
if __name__ == "__main__":
    test_calculate_sum()
````

## Running Scripts vs. Modules

```bash
# These behave DIFFERENTLY!
python mypackage/main.py      # Runs as a script
python -m mypackage.main      # Runs as a module
```

**The difference:**
- Script mode: Python adds the script's folder to `sys.path`
- Module mode: Python adds the current directory to `sys.path`

**Why it matters:** Relative imports only work in module mode!

````python
import sys

# sys.path shows where Python looks for modules
print("Python searches for modules in:")
for i, path in enumerate(sys.path[:5]):
    print(f"  {i+1}. {path}")
````

## The `__main__.py` File

This special file lets you run a package with `python -m packagename`:

```
mypackage/
    __init__.py
    __main__.py    # <-- This runs with python -m mypackage
    core.py
```

```python
# __main__.py
from .core import main

if __name__ == "__main__":
    main()
```

## Common AI-Generated Entry Points

### Pattern 1: Simple Script
```python
# main.py
def main():
    # ... code ...

if __name__ == "__main__":
    main()
```

### Pattern 2: CLI with argparse
```python
import argparse

def main():
    parser = argparse.ArgumentParser()
    parser.add_argument("--input", required=True)
    args = parser.parse_args()
    # ... use args.input ...

if __name__ == "__main__":
    main()
```

### Pattern 3: Async Entry Point
```python
import asyncio

async def main():
    # ... async code ...

if __name__ == "__main__":
    asyncio.run(main())
```

````python
# Quick check: What runs and what doesn't?

print("1. This always runs (top-level code)")

def some_function():
    print("2. This only runs if called")

class SomeClass:
    print("3. This runs when class is defined (at import time!)")

if __name__ == "__main__":
    print("4. This only runs if file is executed directly")
    some_function()
````

## Summary

| Concept | Meaning |
|---------|--------|
| `__name__` | Special variable set by Python |
| `"__main__"` | Value when file is run directly |
| `if __name__ == "__main__":` | Code that only runs when executed, not imported |
| `__main__.py` | Enables `python -m package` |
| `python -m` | Run as module (enables relative imports) |
