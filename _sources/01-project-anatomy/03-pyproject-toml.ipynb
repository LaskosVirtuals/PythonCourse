# Understanding pyproject.toml

Modern Python projects use `pyproject.toml` for configuration. When AI generates a project, you'll often see this file. Let's decode it.

## What Is pyproject.toml?

A single configuration file that replaces:
- `setup.py` - package installation
- `setup.cfg` - package metadata
- `requirements.txt` - dependencies (partially)
- Tool-specific configs (pytest.ini, .flake8, etc.)

**Format:** TOML (Tom's Obvious Minimal Language) - like INI but better.

## Anatomy of pyproject.toml

```toml
# Section 1: Build system (required for packaging)
[build-system]
requires = ["setuptools>=61.0"]
build-backend = "setuptools.build_meta"

# Section 2: Project metadata
[project]
name = "my-package"
version = "1.0.0"
description = "A helpful description"
authors = [{name = "Your Name", email = "you@example.com"}]
readme = "README.md"
requires-python = ">=3.8"
dependencies = [
    "requests>=2.28.0",
    "click>=8.0.0",
]

# Section 3: Optional dependencies
[project.optional-dependencies]
dev = ["pytest", "black", "mypy"]
docs = ["sphinx", "myst-parser"]

# Section 4: CLI entry points
[project.scripts]
mycli = "mypackage.cli:main"

# Section 5: Tool configurations
[tool.pytest.ini_options]
testpaths = ["tests"]

[tool.black]
line-length = 88
```

## Section by Section

### [build-system]

Tells Python how to build your package:

```toml
[build-system]
requires = ["setuptools>=61.0"]  # Build dependencies
build-backend = "setuptools.build_meta"  # Build tool
```

**Common backends:**
- `setuptools.build_meta` - Traditional, widely supported
- `hatchling` - Modern, by PyPA
- `poetry.core.masonry.api` - If using Poetry
- `flit_core.buildapi` - Minimal, for pure Python

### [project]

Core metadata about your package:

```toml
[project]
name = "my-package"           # pip install my-package
version = "1.0.0"             # Semantic versioning
description = "Short desc"    # One-liner
requires-python = ">=3.8"     # Python version requirement
dependencies = [              # Required packages
    "requests>=2.28.0",
    "pydantic>=2.0,<3.0",
]
```

**Version specifiers:**
- `>=2.28.0` - At least this version
- `>=2.0,<3.0` - Range (2.x only)
- `~=2.28` - Compatible release (2.28.x)
- `==2.28.0` - Exact version (avoid if possible)

### [project.scripts]

Creates command-line commands:

```toml
[project.scripts]
mycli = "mypackage.cli:main"
```

After `pip install`, you can run `mycli` from anywhere!

The format is: `command = "module.path:function"`

````python
# Example: What [project.scripts] creates

# If pyproject.toml has:
# [project.scripts]
# greet = "mypackage.cli:main"

# And mypackage/cli.py has:
def main():
    """This function runs when you type 'greet' in terminal."""
    print("Hello from the CLI!")

# After pip install, typing 'greet' runs main()
main()  # Simulate it
````

### [project.optional-dependencies]

Extra dependencies for specific use cases:

```toml
[project.optional-dependencies]
dev = ["pytest", "black", "mypy"]
docs = ["sphinx"]
all = ["mypackage[dev,docs]"]  # Combine groups
```

Install with:
```bash
pip install mypackage[dev]      # Include dev dependencies
pip install mypackage[dev,docs] # Multiple groups
pip install -e ".[dev]"         # Editable with dev deps
```

### [tool.*] Sections

Configuration for various tools:

```toml
# Pytest configuration
[tool.pytest.ini_options]
testpaths = ["tests"]
python_files = "test_*.py"

# Black (code formatter)
[tool.black]
line-length = 88
target-version = ["py38", "py39", "py310"]

# Mypy (type checker)
[tool.mypy]
python_version = "3.10"
strict = true

# Ruff (fast linter)
[tool.ruff]
line-length = 88
select = ["E", "F", "I"]
```

## pyproject.toml vs requirements.txt

| File | Purpose | When to Use |
|------|---------|-------------|
| `pyproject.toml` | Package dependencies | Building/distributing packages |
| `requirements.txt` | Environment reproduction | Deploying applications |

**Best practice:** Use both!
- `pyproject.toml` for package metadata and loose dependencies
- `requirements.txt` (or `requirements-lock.txt`) for pinned versions

## Real-World Example

A complete `pyproject.toml` for a CLI tool:

```toml
[build-system]
requires = ["setuptools>=61.0"]
build-backend = "setuptools.build_meta"

[project]
name = "weather-cli"
version = "0.1.0"
description = "A CLI tool to check the weather"
readme = "README.md"
requires-python = ">=3.8"
dependencies = [
    "click>=8.0",
    "requests>=2.28",
    "rich>=13.0",
]

[project.optional-dependencies]
dev = [
    "pytest>=7.0",
    "pytest-cov",
    "black",
    "mypy",
]

[project.scripts]
weather = "weather_cli.main:cli"

[tool.pytest.ini_options]
testpaths = ["tests"]
addopts = "--cov=weather_cli"

[tool.black]
line-length = 88

[tool.mypy]
python_version = "3.10"
warn_return_any = true
```

## Summary

| Section | Purpose |
|---------|--------|
| `[build-system]` | How to build the package |
| `[project]` | Package name, version, dependencies |
| `[project.scripts]` | CLI commands |
| `[project.optional-dependencies]` | Extra install groups |
| `[tool.*]` | Tool-specific config |
