# Exception Handling

Errors happen. Python's exception system lets you handle them gracefully.

## Basic Try/Except

````python
# Without handling - crashes
# result = 10 / 0  # ZeroDivisionError!

# With handling
try:
    result = 10 / 0
except ZeroDivisionError:
    print("Cannot divide by zero!")
    result = None

print(f"Result: {result}")
````

````python
# Catching the exception object
try:
    number = int("not a number")
except ValueError as e:
    print(f"Error: {e}")
    print(f"Error type: {type(e).__name__}")
````

## Multiple Exception Types

````python
def safe_divide(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        print("Cannot divide by zero")
        return None
    except TypeError:
        print("Invalid types for division")
        return None

print(safe_divide(10, 2))    # Works
print(safe_divide(10, 0))    # ZeroDivisionError
print(safe_divide(10, "a"))  # TypeError
````

````python
# Handle multiple types the same way
try:
    value = int("abc")
except (ValueError, TypeError) as e:
    print(f"Conversion failed: {e}")
````

## The Full Try Statement

````python
def process_file_demo(should_fail=False):
    try:
        print("1. Trying to process...")
        if should_fail:
            raise ValueError("Something went wrong!")
        result = "Success!"
    except ValueError as e:
        print(f"2. Caught error: {e}")
        result = "Failed"
    else:
        # Runs only if NO exception occurred
        print("3. No errors - else block runs")
    finally:
        # ALWAYS runs, even if exception or return
        print("4. Finally block - cleanup here")
    
    return result

print("--- Success case ---")
print(process_file_demo(should_fail=False))

print("\n--- Failure case ---")
print(process_file_demo(should_fail=True))
````

## Common Exception Types

````python
# ValueError - wrong value for operation
try:
    int("not a number")
except ValueError as e:
    print(f"ValueError: {e}")

# TypeError - wrong type for operation
try:
    "hello" + 123
except TypeError as e:
    print(f"TypeError: {e}")

# KeyError - key not in dict
try:
    d = {"a": 1}
    value = d["b"]
except KeyError as e:
    print(f"KeyError: {e}")

# IndexError - index out of range
try:
    lst = [1, 2, 3]
    value = lst[10]
except IndexError as e:
    print(f"IndexError: {e}")

# AttributeError - attribute doesn't exist
try:
    "hello".nonexistent_method()
except AttributeError as e:
    print(f"AttributeError: {e}")
````

## Raising Exceptions

````python
def validate_age(age):
    if not isinstance(age, int):
        raise TypeError(f"Age must be int, got {type(age).__name__}")
    if age < 0:
        raise ValueError("Age cannot be negative")
    if age > 150:
        raise ValueError("Age seems unrealistic")
    return True

try:
    validate_age(-5)
except ValueError as e:
    print(f"Validation failed: {e}")
````

````python
# Re-raising exceptions
def process_data(data):
    try:
        result = int(data)
    except ValueError:
        print("Logging: Failed to convert data")
        raise  # Re-raises the same exception

try:
    process_data("not a number")
except ValueError:
    print("Caught re-raised exception")
````

## Custom Exceptions

````python
# Define custom exception
class ValidationError(Exception):
    """Raised when validation fails."""
    pass

class AuthenticationError(Exception):
    """Raised when authentication fails."""
    def __init__(self, username, message="Authentication failed"):
        self.username = username
        self.message = message
        super().__init__(f"{message} for user '{username}'")

# Using custom exceptions
def login(username, password):
    if not username:
        raise ValidationError("Username required")
    if password != "secret":
        raise AuthenticationError(username, "Invalid password")
    return True

try:
    login("alice", "wrong")
except AuthenticationError as e:
    print(f"Login failed: {e}")
    print(f"Username was: {e.username}")
````

## Exception Chaining

````python
class ConfigError(Exception):
    pass

def load_config(filename):
    try:
        # Simulate file error
        raise FileNotFoundError(f"File '{filename}' not found")
    except FileNotFoundError as e:
        # Chain to new exception
        raise ConfigError(f"Could not load config") from e

try:
    load_config("settings.json")
except ConfigError as e:
    print(f"Error: {e}")
    print(f"Caused by: {e.__cause__}")
````

## AI Code Patterns: Error Handling

````python
# Pattern 1: EAFP - Easier to Ask Forgiveness than Permission
# (Pythonic way)

user_data = {"name": "Alice", "age": 30}

# EAFP (Pythonic)
try:
    email = user_data["email"]
except KeyError:
    email = "no-email@example.com"

# vs LBYL - Look Before You Leap (less Pythonic)
if "email" in user_data:
    email = user_data["email"]
else:
    email = "no-email@example.com"

print(f"Email: {email}")
````

````python
# Pattern 2: Default values with .get()
# (Better than try/except for dict access)

user = {"name": "Alice"}

# Clean way to handle missing keys
email = user.get("email", "unknown")
age = user.get("age")  # Returns None if missing

print(f"Email: {email}")
print(f"Age: {age}")
````

````python
# Pattern 3: Catching specific vs generic exceptions

def risky_operation(data):
    # BAD: Catches everything, hides bugs
    try:
        return process(data)
    except Exception:
        return None

def safer_operation(data):
    # BETTER: Catch specific exceptions
    try:
        return int(data)
    except (ValueError, TypeError) as e:
        print(f"Conversion failed: {e}")
        return None

print(safer_operation("42"))
print(safer_operation("not a number"))
````

````python
# Pattern 4: suppress() - ignore specific exceptions
from contextlib import suppress

# Instead of:
try:
    value = int("abc")
except ValueError:
    pass

# You can use:
with suppress(ValueError):
    value = int("abc")

print("Continued without crash")
````

## Exception Hierarchy

```
├── SystemExit
├── KeyboardInterrupt
├── GeneratorExit
└── Exception
    ├── StopIteration
    ├── ArithmeticError
    │   └── ZeroDivisionError
    ├── LookupError
    │   ├── IndexError
    │   └── KeyError
    ├── OSError
    │   └── FileNotFoundError
    ├── ValueError
    ├── TypeError
    └── RuntimeError
```

````python
# Catching parent catches children
try:
    d = {}
    value = d["missing"]
except LookupError:  # Catches KeyError and IndexError
    print("Lookup failed (could be key or index)")
````

## Summary

| Syntax | Purpose |
|--------|--------|
| `try:` | Code that might fail |
| `except Error:` | Handle specific error |
| `except Error as e:` | Capture error object |
| `else:` | Runs if no exception |
| `finally:` | Always runs (cleanup) |
| `raise Error()` | Throw exception |
| `raise ... from e` | Chain exceptions |
