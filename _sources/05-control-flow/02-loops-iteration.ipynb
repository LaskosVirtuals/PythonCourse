# Loops and Iteration

Python has elegant iteration patterns. Understanding them helps you read AI code fluently.

## For Loops - The Pythonic Way

```python
# Iterate over items directly (not indices!)
fruits = ["apple", "banana", "cherry"]

# Pythonic:
for fruit in fruits:
    print(fruit)

# NOT Pythonic (but you'll see it):
# for i in range(len(fruits)):
#     print(fruits[i])
```

```python
# Iterate over dictionaries
user = {"name": "Alice", "age": 30, "role": "admin"}

# Keys (default)
print("Keys:")
for key in user:
    print(f"  {key}")

# Values
print("\nValues:")
for value in user.values():
    print(f"  {value}")

# Both
print("\nItems:")
for key, value in user.items():
    print(f"  {key}: {value}")
```

## `enumerate` - When You Need the Index

```python
fruits = ["apple", "banana", "cherry"]

# Get index and value together
for index, fruit in enumerate(fruits):
    print(f"{index}: {fruit}")

print()

# Start from different number
for num, fruit in enumerate(fruits, start=1):
    print(f"{num}. {fruit}")
```

## `zip` - Iterate Multiple Sequences

```python
names = ["Alice", "Bob", "Charlie"]
ages = [30, 25, 35]
roles = ["admin", "user", "user"]

# Combine multiple lists
for name, age, role in zip(names, ages, roles):
    print(f"{name} ({age}) - {role}")
```

```python
# zip stops at shortest sequence
short = [1, 2]
long = ["a", "b", "c", "d"]

print("Regular zip:")
for x, y in zip(short, long):
    print(f"  {x}, {y}")

# Use zip_longest to include all
from itertools import zip_longest

print("\nzip_longest:")
for x, y in zip_longest(short, long, fillvalue="?"):
    print(f"  {x}, {y}")
```

## `range` - Generate Number Sequences

```python
# range(stop) - 0 to stop-1
print("range(5):", list(range(5)))

# range(start, stop)
print("range(2, 5):", list(range(2, 5)))

# range(start, stop, step)
print("range(0, 10, 2):", list(range(0, 10, 2)))

# Negative step for countdown
print("range(5, 0, -1):", list(range(5, 0, -1)))
```

## While Loops

```python
# Basic while
count = 0
while count < 5:
    print(f"Count: {count}")
    count += 1
```

```python
# Common pattern: infinite loop with break
import random

attempts = 0
while True:
    attempts += 1
    value = random.randint(1, 10)
    print(f"Attempt {attempts}: got {value}")
    if value == 7:
        print("Found it!")
        break
    if attempts >= 20:  # Safety limit
        print("Gave up")
        break
```

## Loop Control: `break`, `continue`, `else`

```python
# break - exit loop immediately
for i in range(10):
    if i == 5:
        break
    print(i, end=" ")
print("\nLoop ended")
```

```python
# continue - skip to next iteration
for i in range(10):
    if i % 2 == 0:  # Skip even numbers
        continue
    print(i, end=" ")
print()
```

```python
# else on loop - runs if loop completes without break
# (Confusing but you'll see it in AI code!)

def find_item(items, target):
    for item in items:
        if item == target:
            print(f"Found {target}!")
            break
    else:
        # This runs only if break was NOT hit
        print(f"{target} not found")

find_item([1, 2, 3], 2)  # Found
find_item([1, 2, 3], 5)  # Not found
```

## List Comprehensions - Compact Loops

```python
# Basic comprehension
numbers = [1, 2, 3, 4, 5]

# Traditional loop:
squares_loop = []
for n in numbers:
    squares_loop.append(n ** 2)

# Comprehension:
squares_comp = [n ** 2 for n in numbers]

print(f"Loop result: {squares_loop}")
print(f"Comp result: {squares_comp}")
```

```python
# With condition
numbers = range(10)

# Only even numbers, squared
even_squares = [n ** 2 for n in numbers if n % 2 == 0]
print(f"Even squares: {even_squares}")
```

```python
# Nested comprehension (creates flat list)
matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]

# Flatten the matrix
flat = [num for row in matrix for num in row]
print(f"Flattened: {flat}")

# Same as:
flat_loop = []
for row in matrix:
    for num in row:
        flat_loop.append(num)
```

## Dict and Set Comprehensions

```python
# Dict comprehension
names = ["alice", "bob", "charlie"]
name_lengths = {name: len(name) for name in names}
print(f"Name lengths: {name_lengths}")

# Swap keys and values
original = {"a": 1, "b": 2, "c": 3}
swapped = {v: k for k, v in original.items()}
print(f"Swapped: {swapped}")
```

```python
# Set comprehension
numbers = [1, 2, 2, 3, 3, 3, 4, 4, 4, 4]
unique_squares = {n ** 2 for n in numbers}
print(f"Unique squares: {unique_squares}")
```

## Generator Expressions - Memory Efficient

```python
# Generator expression (parentheses instead of brackets)
# Doesn't create full list in memory

# List comprehension - all in memory
squares_list = [n ** 2 for n in range(1000)]

# Generator expression - computed on demand
squares_gen = (n ** 2 for n in range(1000))

print(f"List: {type(squares_list)}, size ~{len(squares_list)} items")
print(f"Generator: {type(squares_gen)}")

# Use in functions that consume iterables
total = sum(n ** 2 for n in range(100))  # No extra brackets needed
print(f"Sum of squares: {total}")
```

## Common AI Code Patterns

```python
# Pattern 1: Filter and transform
users = [
    {"name": "Alice", "active": True},
    {"name": "Bob", "active": False},
    {"name": "Charlie", "active": True},
]

active_names = [u["name"] for u in users if u["active"]]
print(f"Active users: {active_names}")
```

```python
# Pattern 2: Build dict from pairs
keys = ["a", "b", "c"]
values = [1, 2, 3]

mapping = dict(zip(keys, values))
print(f"Mapping: {mapping}")
```

```python
# Pattern 3: any() and all()
numbers = [2, 4, 6, 8, 10]

# Check if all match condition
all_even = all(n % 2 == 0 for n in numbers)
print(f"All even: {all_even}")

# Check if any match condition
has_negative = any(n < 0 for n in numbers)
print(f"Has negative: {has_negative}")
```

## Summary

| Pattern | Example |
|---------|--------|
| For each | `for item in items:` |
| With index | `for i, item in enumerate(items):` |
| Multiple lists | `for a, b in zip(list1, list2):` |
| Dict items | `for k, v in dict.items():` |
| List comp | `[x*2 for x in items if x > 0]` |
| Dict comp | `{k: v*2 for k, v in d.items()}` |
| Generator | `(x*2 for x in items)` |
