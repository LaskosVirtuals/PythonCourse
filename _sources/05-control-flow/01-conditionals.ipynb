# Conditionals and Truthiness

Python's if statements are straightforward, but understanding *truthiness* is key to reading AI code.

## Basic If/Elif/Else

```python
score = 85

if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
elif score >= 70:
    grade = "C"
else:
    grade = "F"

print(f"Score {score} = Grade {grade}")
```

## Truthiness - What Counts as True/False?

Python evaluates many things as True or False in boolean context.

```python
# Falsy values (evaluate to False)
falsy_values = [
    False,      # Boolean False
    None,       # None
    0,          # Zero integer
    0.0,        # Zero float
    "",         # Empty string
    [],         # Empty list
    {},         # Empty dict
    set(),      # Empty set
    (),         # Empty tuple
]

print("Falsy values:")
for val in falsy_values:
    print(f"  {repr(val):15} -> bool = {bool(val)}")
```

```python
# Truthy values (everything else)
truthy_values = [
    True,       # Boolean True
    1,          # Non-zero numbers
    -1,         # Negative numbers too
    "hello",    # Non-empty string
    [1, 2],     # Non-empty list
    {"a": 1},   # Non-empty dict
    " ",        # String with space (not empty!)
]

print("Truthy values:")
for val in truthy_values:
    print(f"  {repr(val):15} -> bool = {bool(val)}")
```

## Common AI Code Pattern: Checking for Empty/None

```python
# Pattern 1: Check if list has items
items = [1, 2, 3]

# Pythonic way (uses truthiness)
if items:
    print("List has items")

# Explicit way (also fine)
if len(items) > 0:
    print("List has items")
```

```python
# Pattern 2: Check for None vs empty
def get_data():
    return []  # Could return None or empty list

data = get_data()

# These are DIFFERENT:
if data is None:
    print("No data returned")
elif not data:  # Empty but not None
    print("Empty data returned")
else:
    print(f"Got data: {data}")
```

```python
# Pattern 3: Default value if None/empty
user_input = ""

# Using 'or' for default (common pattern)
name = user_input or "Anonymous"
print(f"Name: {name}")

# Be careful! This treats 0 and "" the same as None
count = 0
result = count or 10  # Gets 10, not 0!
print(f"Count: {result}")
```

## Comparison Operators

```python
# Standard comparisons
a, b = 5, 10

print(f"a == b: {a == b}")   # Equal
print(f"a != b: {a != b}")   # Not equal
print(f"a < b:  {a < b}")    # Less than
print(f"a <= b: {a <= b}")   # Less or equal
print(f"a > b:  {a > b}")    # Greater than
print(f"a >= b: {a >= b}")   # Greater or equal
```

```python
# Chained comparisons (Python special!)
x = 5

# This is valid Python:
if 0 < x < 10:
    print(f"{x} is between 0 and 10")

# Same as:
if 0 < x and x < 10:
    print(f"{x} is between 0 and 10")
```

## Identity vs Equality: `is` vs `==`

```python
# == checks VALUE equality
# is checks IDENTITY (same object in memory)

a = [1, 2, 3]
b = [1, 2, 3]
c = a

print(f"a == b: {a == b}")  # True (same value)
print(f"a is b: {a is b}")  # False (different objects)
print(f"a is c: {a is c}")  # True (same object)
```

```python
# ALWAYS use 'is' for None, True, False
value = None

# Correct:
if value is None:
    print("Value is None")

# Also correct:
if value is not None:
    print("Value exists")

# Wrong (works but bad style):
# if value == None:
```

## Membership: `in` Operator

```python
# Check if item is in collection
fruits = ["apple", "banana", "cherry"]
settings = {"debug": True, "verbose": False}

print(f"'apple' in fruits: {'apple' in fruits}")
print(f"'grape' in fruits: {'grape' in fruits}")
print(f"'debug' in settings: {'debug' in settings}")  # Checks keys

# Works with strings too
text = "Hello, World!"
print(f"'World' in text: {'World' in text}")
```

## Ternary Expression (Inline If)

```python
# Syntax: value_if_true if condition else value_if_false
age = 20

status = "adult" if age >= 18 else "minor"
print(f"Status: {status}")

# Common in AI code for defaults
config = None
settings = config if config is not None else {}
print(f"Settings: {settings}")
```

```python
# Can be nested (but don't overdo it!)
score = 75

# Readable for simple cases
result = "pass" if score >= 60 else "fail"

# Gets confusing with nesting
grade = "A" if score >= 90 else "B" if score >= 80 else "C" if score >= 70 else "F"
print(f"Grade: {grade}")

# Better to use if/elif for complex logic
```

## Match Statement (Python 3.10+)

Pattern matching - like a powerful switch statement.

```python
# Basic match
def http_status(code):
    match code:
        case 200:
            return "OK"
        case 404:
            return "Not Found"
        case 500:
            return "Server Error"
        case _:  # Default case (underscore = wildcard)
            return "Unknown"

print(http_status(200))
print(http_status(404))
print(http_status(999))
```

```python
# Pattern matching with structure
def process_command(command):
    match command.split():
        case ["quit"]:
            return "Goodbye!"
        case ["hello", name]:
            return f"Hello, {name}!"
        case ["add", *numbers]:
            return f"Sum: {sum(int(n) for n in numbers)}"
        case _:
            return "Unknown command"

print(process_command("quit"))
print(process_command("hello Alice"))
print(process_command("add 1 2 3 4"))
print(process_command("unknown stuff"))
```

## Summary
