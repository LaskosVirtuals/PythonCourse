# Function Signatures

Python has many ways to define function parameters. Let's decode them all.

## The Complete Signature

```python
def function(
    positional_only,    # Before /
    /,                  # Marker: everything before is positional-only
    standard,           # Can be positional or keyword
    *args,              # Capture extra positional args
    keyword_only,       # After * or *args - keyword only
    **kwargs            # Capture extra keyword args
) -> ReturnType:
```

## Basic Parameters

````python
# Simple function with positional parameters
def greet(name, greeting="Hello"):
    return f"{greeting}, {name}!"

# All these work:
print(greet("Alice"))                    # Positional
print(greet("Bob", "Hi"))               # Both positional
print(greet(name="Charlie"))            # Keyword
print(greet(greeting="Hey", name="Dave")) # Keywords (any order)
````

## `*args` - Variable Positional Arguments

````python
def sum_all(*numbers):
    """Accept any number of positional arguments."""
    print(f"Received: {numbers}")  # It's a tuple!
    return sum(numbers)

print(sum_all(1, 2))           # 3
print(sum_all(1, 2, 3, 4, 5))  # 15
print(sum_all())               # 0
````

## `**kwargs` - Variable Keyword Arguments

````python
def create_user(**kwargs):
    """Accept any keyword arguments."""
    print(f"Received: {kwargs}")  # It's a dict!
    return kwargs

user = create_user(name="Alice", age=30, role="admin")
print(f"User: {user}")
````

## Combining `*args` and `**kwargs`

````python
def flexible(required, *args, **kwargs):
    """Common pattern - pass through to another function."""
    print(f"required: {required}")
    print(f"args: {args}")
    print(f"kwargs: {kwargs}")

flexible("first", 2, 3, 4, name="Alice", debug=True)
````

## Keyword-Only Arguments (after `*`)

````python
# The * forces everything after it to be keyword-only
def fetch_data(url, *, timeout=30, retry=True, headers=None):
    """timeout, retry, headers MUST be passed as keywords."""
    print(f"Fetching {url}")
    print(f"  timeout={timeout}, retry={retry}")

# These work:
fetch_data("https://api.com")                      # Defaults
fetch_data("https://api.com", timeout=60)         # Keyword
fetch_data("https://api.com", retry=False, timeout=10)  # Keywords

# This fails:
# fetch_data("https://api.com", 60)  # TypeError!
````

## Positional-Only Arguments (before `/`)

Python 3.8+ feature. Forces arguments to be positional.

````python
def divide(x, y, /):
    """x and y MUST be positional."""
    return x / y

# Works:
print(divide(10, 2))  # 5.0

# Fails:
# print(divide(x=10, y=2))  # TypeError!
````

## Full Example: All Parameter Types

````python
def complex_function(
    pos_only,           # Must be positional
    /,
    standard,           # Can be either
    *args,              # Extra positional
    kw_only,            # Must be keyword
    **kwargs            # Extra keywords
):
    print(f"pos_only: {pos_only}")
    print(f"standard: {standard}")
    print(f"args: {args}")
    print(f"kw_only: {kw_only}")
    print(f"kwargs: {kwargs}")

complex_function(
    "positional",      # pos_only
    "either",          # standard (as positional)
    "extra1", "extra2", # *args
    kw_only="required", # kw_only
    extra_key="value"   # **kwargs
)
````

## Unpacking Arguments

````python
def point_distance(x1, y1, x2, y2):
    return ((x2-x1)**2 + (y2-y1)**2) ** 0.5

# Unpack a list/tuple with *
coords = [0, 0, 3, 4]
print(point_distance(*coords))  # Same as point_distance(0, 0, 3, 4)

# Unpack a dict with **
point_dict = {"x1": 0, "y1": 0, "x2": 3, "y2": 4}
print(point_distance(**point_dict))  # Same as point_distance(x1=0, y1=0, x2=3, y2=4)
````

## Common AI Code Patterns

````python
# Pattern 1: Wrapper function passing through args
def logged_call(func, *args, **kwargs):
    print(f"Calling {func.__name__}")
    result = func(*args, **kwargs)
    print(f"Result: {result}")
    return result

def add(a, b):
    return a + b

logged_call(add, 2, 3)
````

````python
# Pattern 2: Config with keyword-only for clarity
def connect(
    host: str,
    port: int,
    *,
    timeout: int = 30,
    ssl: bool = True,
    verify: bool = True
):
    """Connection settings are keyword-only for clarity."""
    print(f"Connecting to {host}:{port}")
    print(f"  SSL: {ssl}, Verify: {verify}, Timeout: {timeout}")

# Clear what each argument means
connect("localhost", 5432, ssl=False, timeout=60)
````

## Summary

| Syntax | Meaning |
|--------|--------|
| `arg` | Standard argument |
| `arg=default` | Default value |
| `*args` | Capture extra positional |
| `**kwargs` | Capture extra keyword |
| `*` alone | Everything after is keyword-only |
| `/` | Everything before is positional-only |
| `*list` | Unpack list as positional args |
| `**dict` | Unpack dict as keyword args |
