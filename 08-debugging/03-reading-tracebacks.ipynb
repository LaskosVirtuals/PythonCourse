# Reading Tracebacks

Python's error messages tell you exactly what went wrong - if you know how to read them.

## Anatomy of a Traceback

````python
# Let's create an error and examine the traceback
def level_3():
    return 1 / 0  # This will fail

def level_2():
    return level_3()

def level_1():
    return level_2()

try:
    level_1()
except ZeroDivisionError:
    import traceback
    traceback.print_exc()
````

## Reading Order: Bottom to Top

```
Traceback (most recent call last):     # Header
  File "script.py", line 10, in level_1  # Oldest call
    return level_2()
  File "script.py", line 7, in level_2   # Middle call
    return level_3()
  File "script.py", line 4, in level_3   # Most recent call
    return 1 / 0
ZeroDivisionError: division by zero      # Error type and message
```

**Key insight:** The actual error is at the BOTTOM. Read from bottom to top to understand the call chain.

## Common Error Types

````python
# 1. SyntaxError - Code structure is wrong
# (Can't demonstrate in running code - it won't run!)

# Common causes:
# - Missing colon after if/for/def
# - Mismatched parentheses/brackets
# - Invalid indentation

print("SyntaxError examples:")
print("  if x == 5    # Missing colon")
print("  def func()   # Missing colon")
print("  print('hello'  # Missing closing paren")
````

````python
# 2. NameError - Variable/function doesn't exist
try:
    print(undefined_variable)
except NameError as e:
    print(f"NameError: {e}")
    print("Fix: Check spelling, ensure variable is defined before use")
````

````python
# 3. TypeError - Wrong type for operation
try:
    result = "hello" + 5
except TypeError as e:
    print(f"TypeError: {e}")
    print("Fix: Convert types explicitly (str(5) or int('5'))")
````

````python
# 4. ValueError - Right type, wrong value
try:
    number = int("not a number")
except ValueError as e:
    print(f"ValueError: {e}")
    print("Fix: Validate input before converting")
````

````python
# 5. KeyError - Key not in dictionary
try:
    data = {"name": "Alice"}
    age = data["age"]
except KeyError as e:
    print(f"KeyError: {e}")
    print("Fix: Use .get() with default or check 'key in dict' first")
````

````python
# 6. IndexError - Index out of range
try:
    items = [1, 2, 3]
    item = items[10]
except IndexError as e:
    print(f"IndexError: {e}")
    print(f"Fix: Check len() first. List has {len(items)} items.")
````

````python
# 7. AttributeError - Object doesn't have that attribute
try:
    text = "hello"
    text.append("world")  # Strings don't have append!
except AttributeError as e:
    print(f"AttributeError: {e}")
    print("Fix: Check object type and available methods")
````

````python
# 8. ImportError / ModuleNotFoundError
try:
    from nonexistent_module import something
except ModuleNotFoundError as e:
    print(f"ModuleNotFoundError: {e}")
    print("Fix: Check module name, install package if needed")
````

````python
# 9. FileNotFoundError
try:
    with open("nonexistent_file.txt") as f:
        content = f.read()
except FileNotFoundError as e:
    print(f"FileNotFoundError: {e}")
    print("Fix: Check file path, working directory, file exists")
````

## Decoding Complex Tracebacks

````python
# Traceback from library code
import json

try:
    data = json.loads('{invalid json}')
except json.JSONDecodeError as e:
    import traceback
    traceback.print_exc()
````

**When tracebacks show library code:**
1. Skip library frames (in site-packages)
2. Find your code - that's where the bug is
3. The library error tells you what went wrong
4. Your code tells you where you made the mistake

````python
# Chained exceptions (Python 3)
def process_config():
    try:
        data = {"port": "not a number"}
        port = int(data["port"])
    except ValueError as e:
        raise RuntimeError("Invalid config") from e

try:
    process_config()
except RuntimeError:
    import traceback
    traceback.print_exc()
````

## Traceback Quick Reference

| Error | Common Cause |
|-------|-------------|
| `SyntaxError` | Typo in code structure |
| `NameError` | Undefined variable |
| `TypeError` | Wrong type for operation |
| `ValueError` | Invalid value for operation |
| `KeyError` | Missing dictionary key |
| `IndexError` | List index out of range |
| `AttributeError` | Object missing attribute |
| `ImportError` | Module not found |
| `FileNotFoundError` | File doesn't exist |

## Real-World Example

````python
# Common AI-generated code pattern with potential bug
def fetch_user_data(users, user_id):
    user = users[user_id]  # Potential KeyError
    return {
        'name': user['name'],
        'email': user['profile']['email'],  # Potential KeyError (nested)
        'age': int(user['age']),  # Potential ValueError
    }

# Test with problematic data
users = {
    '1': {'name': 'Alice', 'profile': {}, 'age': '30'},  # Missing email!
}

try:
    result = fetch_user_data(users, '1')
except KeyError as e:
    print(f"KeyError: {e}")
    print("The nested 'email' key is missing from profile")
````

````python
# Fixed version with defensive coding
def fetch_user_data_safe(users, user_id):
    user = users.get(user_id)
    if user is None:
        return None
    
    profile = user.get('profile', {})
    
    # Handle age conversion safely
    try:
        age = int(user.get('age', 0))
    except (ValueError, TypeError):
        age = 0
    
    return {
        'name': user.get('name', 'Unknown'),
        'email': profile.get('email', 'no-email'),
        'age': age,
    }

result = fetch_user_data_safe(users, '1')
print(f"Safe result: {result}")
````

## Summary: Reading Tracebacks

1. **Start at the bottom** - Error type and message
2. **Work up** - Find your code (not library code)
3. **Look at the line** - What operation failed?
4. **Check variables** - What were the actual values?
5. **Fix at the source** - Add validation/error handling
