# Testing

Catch bugs before they bite. Testing is essential when working with AI-generated code.

## Why Test AI Code?

1. **Verify correctness** - Does it actually work?
2. **Catch edge cases** - AI often misses these
3. **Document behavior** - Tests show how code should work
4. **Safe refactoring** - Change code confidently

## Simple Assertion Tests

```python
# Function to test
def add(a, b):
    return a + b

# Simple assertion tests
assert add(2, 3) == 5, "Basic addition failed"
assert add(-1, 1) == 0, "Negative numbers failed"
assert add(0, 0) == 0, "Zeros failed"

print("All tests passed!")
```

```python
# Testing edge cases
def divide(a, b):
    if b == 0:
        raise ValueError("Cannot divide by zero")
    return a / b

# Test normal cases
assert divide(10, 2) == 5.0
assert divide(-6, 2) == -3.0

# Test edge case (expecting error)
try:
    divide(5, 0)
    assert False, "Should have raised ValueError"
except ValueError:
    pass  # Expected!

print("Division tests passed!")
```

## Using pytest (The Standard)

```python
# pytest conventions:
# - Test files: test_*.py or *_test.py
# - Test functions: test_*
# - Test classes: Test*

# Example test file: test_calculator.py

def test_add_positive_numbers():
    """Test adding two positive numbers."""
    result = add(2, 3)
    assert result == 5

def test_add_negative_numbers():
    """Test adding negative numbers."""
    result = add(-1, -1)
    assert result == -2

def test_add_mixed():
    """Test adding positive and negative."""
    result = add(-1, 2)
    assert result == 1

# Run with: pytest test_calculator.py -v
print("pytest example tests defined")
```

## Testing for Exceptions

```python
# With pytest, use pytest.raises
# import pytest
#
# def test_divide_by_zero():
#     with pytest.raises(ValueError):
#         divide(5, 0)
#
# def test_divide_by_zero_message():
#     with pytest.raises(ValueError, match="Cannot divide"):
#         divide(5, 0)

# Without pytest:
def test_exception_raised():
    try:
        divide(5, 0)
        raise AssertionError("Expected ValueError")
    except ValueError as e:
        assert "zero" in str(e).lower()

test_exception_raised()
print("Exception test passed!")
```

## Test Fixtures

```python
# Fixtures provide test data
# With pytest:
#
# @pytest.fixture
# def sample_users():
#     return [
#         {"name": "Alice", "age": 30},
#         {"name": "Bob", "age": 25},
#     ]
#
# def test_count_users(sample_users):
#     assert len(sample_users) == 2

# Without pytest - use setup functions
def get_sample_users():
    """Test fixture."""
    return [
        {"name": "Alice", "age": 30},
        {"name": "Bob", "age": 25},
    ]

def test_user_names():
    users = get_sample_users()
    names = [u["name"] for u in users]
    assert "Alice" in names
    assert "Bob" in names

test_user_names()
print("Fixture test passed!")
```

## Testing AI-Generated Code Patterns

```python
# Pattern 1: Test with various input types
def process_data(data):
    """AI-generated function to test."""
    if data is None:
        return []
    return [str(item).upper() for item in data]

# Test various inputs
def test_process_data():
    # Normal case
    assert process_data(["a", "b"]) == ["A", "B"]
    
    # Empty input
    assert process_data([]) == []
    
    # None input
    assert process_data(None) == []
    
    # Mixed types
    assert process_data([1, "a", 2.5]) == ["1", "A", "2.5"]

test_process_data()
print("Process data tests passed!")
```

```python
# Pattern 2: Test boundary conditions
def paginate(items, page, per_page):
    """Return items for given page."""
    if page < 1:
        page = 1
    start = (page - 1) * per_page
    end = start + per_page
    return items[start:end]

def test_pagination():
    items = list(range(25))  # 0-24
    
    # First page
    assert paginate(items, 1, 10) == list(range(10))
    
    # Middle page
    assert paginate(items, 2, 10) == list(range(10, 20))
    
    # Last page (partial)
    assert paginate(items, 3, 10) == list(range(20, 25))
    
    # Beyond last page
    assert paginate(items, 10, 10) == []
    
    # Invalid page (should handle gracefully)
    assert paginate(items, 0, 10) == list(range(10))
    assert paginate(items, -1, 10) == list(range(10))

test_pagination()
print("Pagination tests passed!")
```

```python
# Pattern 3: Test error handling
def validate_email(email):
    """Simple email validation."""
    if not email or not isinstance(email, str):
        raise ValueError("Email must be a non-empty string")
    if "@" not in email:
        raise ValueError("Email must contain @")
    if "." not in email.split("@")[-1]:
        raise ValueError("Email domain must contain .")
    return True

def test_email_validation():
    # Valid emails
    assert validate_email("user@example.com") == True
    assert validate_email("test@sub.domain.org") == True
    
    # Invalid emails should raise
    invalid_emails = [
        None,           # None
        "",             # Empty
        "notanemail",   # No @
        "bad@domain",   # No . in domain
        123,            # Wrong type
    ]
    
    for email in invalid_emails:
        try:
            validate_email(email)
            print(f"FAILED: {email!r} should have raised")
        except ValueError:
            pass  # Expected

test_email_validation()
print("Email validation tests passed!")
```

## Parametrized Testing

```python
# With pytest:
# @pytest.mark.parametrize("input,expected", [
#     ("hello", "HELLO"),
#     ("World", "WORLD"),
#     ("", ""),
# ])
# def test_upper(input, expected):
#     assert input.upper() == expected

# Without pytest - loop over test cases
def test_upper_parametrized():
    test_cases = [
        ("hello", "HELLO"),
        ("World", "WORLD"),
        ("", ""),
        ("123", "123"),
        ("MiXeD", "MIXED"),
    ]
    
    for input_val, expected in test_cases:
        result = input_val.upper()
        assert result == expected, f"Failed for {input_val!r}"
        print(f"  {input_val!r} -> {result!r} ✓")

test_upper_parametrized()
```

## Test-Driven Development (TDD) for AI Code

```python
# Step 1: Write the test FIRST (before asking AI for code)
def test_calculate_discount():
    """Tests for a discount calculator."""
    # Regular price, no discount
    assert calculate_discount(100, 0) == 100
    
    # 10% discount
    assert calculate_discount(100, 10) == 90
    
    # 50% discount
    assert calculate_discount(100, 50) == 50
    
    # Edge: 100% discount
    assert calculate_discount(100, 100) == 0

# Step 2: Ask AI to implement (or write yourself)
def calculate_discount(price, discount_percent):
    """Calculate price after discount."""
    if discount_percent < 0 or discount_percent > 100:
        raise ValueError("Discount must be 0-100")
    return price * (1 - discount_percent / 100)

# Step 3: Run tests
test_calculate_discount()
print("TDD tests passed!")
```

## Quick Testing Checklist

When testing AI-generated code:

1. **Happy path** - Normal expected input
2. **Empty input** - `[]`, `""`, `{}`, `None`
3. **Edge cases** - Boundaries, zeros, negatives
4. **Wrong types** - What if string instead of int?
5. **Large input** - Performance issues?
6. **Error cases** - Does it fail gracefully?
