# Packages and Modules

When AI generates code, you'll see imports like:

```python
from myproject.utils.helpers import process_data
```

What do all these dots mean? How does Python find these files? What's `__init__.py`?

This lesson demystifies Python's package structure.

## Modules vs. Packages

| Term | What It Is | Example |
|------|-----------|--------|
| **Module** | A single `.py` file | `utils.py` |
| **Package** | A folder containing modules | `myproject/` folder |

**Key insight:** When you see dots in imports, you're navigating folders.

## How Import Paths Map to Files

```python
# This import:
from myproject.utils.helpers import process_data

# Maps to this file structure:
myproject/
    __init__.py
    utils/
        __init__.py
        helpers.py      <-- This file!
            def process_data():  <-- This function!
```

Each dot (`.`) represents going into a subfolder.

## The `__init__.py` File

This special file makes a folder into a Python package.

### What It Does

1. **Tells Python "this folder is a package"**
2. **Runs when you import the package**
3. **Controls what gets exported**

````python
# Let's simulate a package structure
# In a real project, these would be files

# Imagine this file structure:
# mypackage/
#     __init__.py
#     core.py
#     helpers.py

# If __init__.py is EMPTY:
#   - from mypackage import core     # Works
#   - from mypackage import helpers  # Works  
#   - from mypackage import MyClass  # Fails! (unless core.py is imported and re-exported)

# If __init__.py contains:
#   from .core import MyClass
#   from .helpers import process_data
#
# Then:
#   - from mypackage import MyClass     # Works!
#   - from mypackage import process_data # Works!

print("See the examples below for how this works in practice.")
````

### Common `__init__.py` Patterns

**Pattern 1: Empty file** (minimal)
```python
# __init__.py
# (empty - just marks folder as package)
```

**Pattern 2: Re-export for convenience**
```python
# __init__.py
from .core import MainClass, helper_function
from .models import User, Product

# Now users can do:
# from mypackage import MainClass
# Instead of:
# from mypackage.core import MainClass
```

**Pattern 3: Define public API with `__all__`**
```python
# __init__.py
from .core import MainClass, _internal_helper

__all__ = ["MainClass"]  # Only MainClass is "public"

# from mypackage import * 
# ^ This only imports MainClass, not _internal_helper
```

## Relative vs. Absolute Imports

When importing within a package, you have two options:

### Absolute Imports

Start from the root package:

```python
# In mypackage/utils/formatter.py
from mypackage.core import MainClass
from mypackage.utils.helpers import process
```

**Pros:** Clear where things come from  
**Cons:** Breaks if you rename the package

### Relative Imports

Use dots to navigate relative to current file:

```python
# In mypackage/utils/formatter.py

from . import helpers          # Same folder (utils/)
from .helpers import process   # Same folder, specific item
from .. import core            # Parent folder (mypackage/)
from ..core import MainClass   # Parent folder, specific item
```

**The dots mean:**
- `.` = current package/folder
- `..` = parent package/folder
- `...` = grandparent (rare)

````python
# Visual example of relative imports:

structure = """
mypackage/
    __init__.py
    core.py              
    utils/
        __init__.py
        helpers.py      <-- If we're HERE
        formatter.py
    models/
        __init__.py
        user.py

From helpers.py:
    from . import formatter     →  utils/formatter.py
    from .. import core         →  mypackage/core.py
    from ..models import user   →  mypackage/models/user.py
"""
print(structure)
````

## Common Package Layouts

AI generates different project structures. Here are the most common:

### Layout 1: Flat (Simple Scripts)

```
myproject/
    main.py
    utils.py
    config.py
```

Import like: `from utils import helper_function`

### Layout 2: Package in Root

```
myproject/
    mypackage/
        __init__.py
        core.py
        utils.py
    tests/
        test_core.py
    setup.py
```

Import like: `from mypackage.core import MainClass`

### Layout 3: `src` Layout (Modern Best Practice)

```
myproject/
    src/
        mypackage/
            __init__.py
            core.py
            utils/
                __init__.py
                helpers.py
    tests/
        test_core.py
    pyproject.toml
```

The `src/` folder prevents accidentally importing the wrong thing during development.

## How Python Finds Modules

When you write `import something`, Python searches in order:

1. **Built-in modules** (like `sys`, `os`)
2. **Current directory** (where you ran Python)
3. **PYTHONPATH** (environment variable)
4. **Installed packages** (`site-packages`)

````python
# See Python's search path
import sys

print("Python searches for modules in these locations:")
for i, path in enumerate(sys.path[:5]):  # First 5 paths
    print(f"  {i+1}. {path}")
print(f"  ... and {len(sys.path)-5} more locations")
````

## AI Coder Gotchas

### Gotcha 1: Running Scripts vs. Modules

```bash
# These behave differently!
python mypackage/main.py        # Runs as script (relative imports may fail)
python -m mypackage.main        # Runs as module (relative imports work)
```

If AI-generated code has relative imports that fail, try running with `-m`.

### Gotcha 2: Name Collision

```python
# If you have a file named random.py in your project:
import random  # This imports YOUR random.py, not Python's!

# Fix: rename your file to something else
```

**Never name your files after standard library modules!**

### Gotcha 3: Missing `__init__.py`

```
mypackage/
    utils/          <-- No __init__.py
        helpers.py

# This fails:
from mypackage.utils import helpers
# ModuleNotFoundError: No module named 'mypackage.utils'
```

**Note:** Python 3.3+ has "namespace packages" that work without `__init__.py`, but most code still uses them for compatibility.

## Quick Check: Trace the Import

````python
# Given this structure:
structure = """
webapp/
    __init__.py
    app.py
    api/
        __init__.py
        routes.py
        handlers/
            __init__.py
            auth.py
            users.py
    models/
        __init__.py
        user.py
"""
print(structure)

# What file does each import refer to?
# 1. from webapp import app
#    Answer: webapp/app.py

# 2. from webapp.api import routes  
#    Answer: webapp/api/routes.py

# 3. from webapp.api.handlers.auth import login
#    Answer: webapp/api/handlers/auth.py, function 'login'
````

````python
# In webapp/api/handlers/auth.py, what do these relative imports mean?

# from . import users
# → webapp/api/handlers/users.py

# from .. import routes
# → webapp/api/routes.py

# from ...models import user
# → webapp/models/user.py

print("Understanding complete! You can now trace any import path.")
````

## Summary

| Concept | Meaning |
|---------|--------|
| Module | Single `.py` file |
| Package | Folder with `__init__.py` |
| Dots in imports | Navigate into folders |
| `__init__.py` | Makes folder a package, controls exports |
| `.` in relative import | Current folder |
| `..` in relative import | Parent folder |
| `__all__` | List of public exports |
