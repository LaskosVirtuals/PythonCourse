# Import Statement Forms

When you look at AI-generated code, you'll see many different ways to write imports. This lesson decodes all of them.

## The Basic Forms

There are essentially **two** import patterns, with variations:

1. `import module` - bring in the whole module
2. `from module import thing` - bring in specific items

## Form 1: `import module`

This imports the entire module. You access its contents using **dot notation**.

````python
# Import the entire 'os' module
import os

# Now use it with dot notation: module.function()
current_directory = os.getcwd()
print(f"Current directory: {current_directory}")

# Everything in 'os' is available via os.something
print(f"Path separator: {os.sep}")
````

### AI Coder Insight

When you see `os.getcwd()` in AI code, the `os.` prefix tells you:
- `getcwd` comes from the `os` module
- There must be an `import os` somewhere above
- If you can't find the import, that's a bug!

## Form 2: `from module import thing`

This imports only specific items. You use them **directly** without the module prefix.

````python
# Import specific functions from os.path
from os.path import join, exists

# Use directly - no prefix needed
my_path = join("folder", "subfolder", "file.txt")
print(f"Path: {my_path}")
print(f"Exists: {exists(my_path)}")

# Note: os.path.join would NOT work here (we didn't import os.path)
# Try uncommenting the line below to see the error:
# print(os.path.join("a", "b"))
````

### AI Coder Insight

When you see a function used without a prefix (like `join(...)` instead of `os.path.join(...)`), look for a `from X import Y` statement at the top of the file.

## Form 3: `import module as alias`

This imports a module with a **shorter name**. Very common with long module names.

````python
# These aliases are SO common that everyone uses them
import datetime as dt

# Now use dt instead of datetime
now = dt.datetime.now()
print(f"Current time: {now}")

# Without the alias, you'd write:
# import datetime
# now = datetime.datetime.now()  # Repetitive!
````

### Common Aliases You'll See in AI Code

These are **universally standard** - AI will always use them:

```python
import numpy as np          # NumPy for numerical computing
import pandas as pd         # Pandas for data analysis
import matplotlib.pyplot as plt  # Matplotlib for plotting
import tensorflow as tf     # TensorFlow for ML
import seaborn as sns       # Seaborn for statistical plots
```

If you see `np.array()` or `pd.DataFrame()`, now you know where they come from!

## Form 4: `from module import thing as alias`

Import a specific item and rename it.

````python
# Rename an import to avoid conflicts or for clarity
from os.path import join as path_join

# Now use the alias
result = path_join("folder", "file.txt")
print(result)

# This is useful when you have multiple 'join' functions
# from different modules
````

## Form 5: `from module import *` (Star Import)

This imports **everything** from a module. You'll see it, but it's generally discouraged.

````python
# Star import - brings in everything
from math import *

# Now all math functions are available directly
print(f"Pi: {pi}")
print(f"Square root of 16: {sqrt(16)}")
print(f"Cosine of 0: {cos(0)}")
````

### Why Star Imports Are Discouraged

```python
from module_a import *
from module_b import *

result = some_function()  # Which module is this from??
```

**Problems:**
- Hard to know where things come from
- Can accidentally overwrite functions with same names
- Makes debugging much harder

**When AI uses them:** Usually in quick scripts or interactive work. In production code, prefer explicit imports.

## Form 6: Importing Multiple Items

You can import several things at once. Python style prefers parentheses for long imports.

````python
# Multiple imports on one line (short list)
from os.path import join, exists, dirname

# Multiple imports with parentheses (long list, cleaner)
from typing import (
    Optional,
    List,
    Dict,
    Union,
    Callable,
)

print("Imports successful!")
print(f"Optional is: {Optional}")
````

## Summary

| Form | Example | How to Use |
|------|---------|------------|
| `import X` | `import os` | `os.getcwd()` |
| `from X import Y` | `from os import getcwd` | `getcwd()` |
| `import X as Z` | `import numpy as np` | `np.array()` |
| `from X import Y as Z` | `from os import getcwd as cwd` | `cwd()` |
| `from X import *` | `from math import *` | `sqrt()`, `sin()`, etc. |
