# Type Hints

Modern Python code is full of type annotations. Let's understand what they mean.

## What Are Type Hints?

Type hints are **documentation** that describe expected types. They don't change how code runs!

```python
# Without type hints
def greet(name):
    return f"Hello, {name}"

# With type hints
def greet_typed(name: str) -> str:
    return f"Hello, {name}"

# Both work exactly the same!
print(greet("World"))
print(greet_typed("World"))

# Python doesn't enforce types at runtime
print(greet_typed(123))  # Works! (even though it says str)
```

## Basic Types

```python
# Built-in types
def process(
    text: str,          # String
    count: int,         # Integer
    ratio: float,       # Float
    active: bool,       # Boolean
    data: bytes,        # Bytes
) -> None:              # Returns nothing
    pass

# Collection types (Python 3.9+)
def work_with_collections(
    items: list,        # Any list
    mapping: dict,      # Any dict
    unique: set,        # Any set
    sequence: tuple,    # Any tuple
) -> None:
    pass
```

## Generic Types (with contents)

```python
# Specify what's IN the collection
# Python 3.9+ syntax:
def process_data(
    numbers: list[int],                    # List of integers
    user_ages: dict[str, int],            # Dict with str keys, int values
    coordinates: tuple[float, float],      # Tuple of exactly 2 floats
    tags: set[str],                        # Set of strings
) -> list[str]:
    return [str(n) for n in numbers]

result = process_data(
    [1, 2, 3],
    {"Alice": 30, "Bob": 25},
    (1.0, 2.0),
    {"python", "coding"}
)
print(result)
```

## Optional and Union Types

```python
from typing import Optional, Union

# Optional = can be None
def find_user(id: int) -> Optional[str]:
    """Returns username or None if not found."""
    users = {1: "Alice", 2: "Bob"}
    return users.get(id)  # Returns None if not found

print(find_user(1))  # "Alice"
print(find_user(99)) # None

# Union = one of several types
def process(value: Union[str, int]) -> str:
    return str(value)

# Python 3.10+ uses | instead:
# def process(value: str | int) -> str:
#     return str(value)
```

## Modern Type Hint Syntax (Python 3.10+)

```python
# Old style (still works)
from typing import Optional, Union, List, Dict

def old_style(
    items: List[int],
    config: Dict[str, str],
    value: Optional[str],
    data: Union[str, int],
) -> None:
    pass

# New style (Python 3.9+ for generics, 3.10+ for |)
def new_style(
    items: list[int],        # lowercase!
    config: dict[str, str],  # lowercase!
    value: str | None,       # | instead of Optional
    data: str | int,         # | instead of Union
) -> None:
    pass
```

## Common Type Imports

```python
from typing import (
    # Basic
    Optional,      # X | None
    Union,         # X | Y
    Any,           # Anything goes
    
    # Collections (use lowercase in 3.9+)
    List, Dict, Set, Tuple,
    
    # Callables
    Callable,      # Function type
    
    # Special
    TypeVar,       # Generic type parameter
    Generic,       # Base for generic classes
    Protocol,      # Structural typing
    Literal,       # Specific values only
    Final,         # Cannot be reassigned
    TypedDict,     # Dict with specific keys
)

print("Common typing imports loaded!")
```

## Callable Type

```python
from typing import Callable

# Callable[[arg_types], return_type]

def apply_twice(
    func: Callable[[int], int],  # Takes int, returns int
    value: int
) -> int:
    return func(func(value))

def double(x: int) -> int:
    return x * 2

result = apply_twice(double, 5)
print(f"Result: {result}")  # 20
```

## TypedDict - Dictionaries with Structure

```python
from typing import TypedDict

class UserDict(TypedDict):
    name: str
    age: int
    email: str

def process_user(user: UserDict) -> str:
    return f"{user['name']} ({user['age']})"

user: UserDict = {
    "name": "Alice",
    "age": 30,
    "email": "alice@example.com"
}

print(process_user(user))
```

## Literal Types - Specific Values

```python
from typing import Literal

def set_mode(mode: Literal["read", "write", "append"]) -> None:
    print(f"Mode set to: {mode}")

set_mode("read")   # OK
set_mode("write")  # OK
# set_mode("delete")  # Type checker would complain!
```

## Type Aliases

```python
from typing import TypeAlias

# Create meaningful names for complex types
UserId: TypeAlias = int
UserData: TypeAlias = dict[str, str | int | None]
Coordinates: TypeAlias = tuple[float, float]

def get_user(user_id: UserId) -> UserData:
    return {"id": user_id, "name": "Alice", "age": 30}

def distance(p1: Coordinates, p2: Coordinates) -> float:
    return ((p2[0]-p1[0])**2 + (p2[1]-p1[1])**2) ** 0.5

print(get_user(123))
print(distance((0, 0), (3, 4)))
```

## Summary

| Syntax | Meaning |
|--------|--------|
| `x: int` | x is an integer |
| `-> str` | Function returns string |
| `list[int]` | List of integers |
| `dict[str, int]` | Dict with str keys, int values |
| `str \| None` | String or None (Optional) |
| `str \| int` | String or integer (Union) |
| `Callable[[int], str]` | Function taking int, returning str |
| `Literal["a", "b"]` | Only "a" or "b" allowed |
| `Any` | Any type |
