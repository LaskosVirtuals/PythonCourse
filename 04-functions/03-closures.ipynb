# Closures and Higher-Order Functions

Functions can return functions. This powerful pattern appears everywhere in Python.

## What Is a Closure?

A closure is a function that remembers values from its enclosing scope.

```python
def make_multiplier(n):
    """Returns a function that multiplies by n."""
    def multiplier(x):
        return x * n  # 'n' is remembered from outer scope!
    return multiplier

# Create specialized functions
double = make_multiplier(2)
triple = make_multiplier(3)

print(double(5))   # 10
print(triple(5))   # 15
print(double(10))  # 20
```

```python
# The 'n' is "closed over" - stored with the function
print(f"double remembers n={double.__closure__[0].cell_contents}")
print(f"triple remembers n={triple.__closure__[0].cell_contents}")
```

## Practical Closure Examples

```python
# Counter factory
def make_counter(start=0):
    count = start
    
    def counter():
        nonlocal count  # Modify outer variable
        count += 1
        return count
    
    return counter

counter_a = make_counter()
counter_b = make_counter(100)

print(counter_a())  # 1
print(counter_a())  # 2
print(counter_b())  # 101
print(counter_a())  # 3 - independent!
```

```python
# Logger factory
def make_logger(prefix):
    def log(message):
        print(f"[{prefix}] {message}")
    return log

info = make_logger("INFO")
error = make_logger("ERROR")

info("Application started")
error("Something went wrong")
```

## Lambda Functions

Anonymous functions for simple operations.

```python
# Lambda syntax: lambda arguments: expression

# These are equivalent:
def add(x, y):
    return x + y

add_lambda = lambda x, y: x + y

print(add(2, 3))        # 5
print(add_lambda(2, 3)) # 5
```

```python
# Common uses of lambda
numbers = [3, 1, 4, 1, 5, 9, 2, 6]

# Sorting with custom key
words = ["banana", "apple", "cherry"]
sorted_by_length = sorted(words, key=lambda x: len(x))
print(f"By length: {sorted_by_length}")

# Filter with lambda
evens = list(filter(lambda x: x % 2 == 0, numbers))
print(f"Evens: {evens}")

# Map with lambda
squared = list(map(lambda x: x**2, numbers))
print(f"Squared: {squared}")
```

## `functools.partial` - Pre-fill Arguments

```python
from functools import partial

def greet(greeting, name, punctuation="!"):
    return f"{greeting}, {name}{punctuation}"

# Create specialized versions
say_hello = partial(greet, "Hello")
say_hi = partial(greet, "Hi", punctuation=".")

print(say_hello("Alice"))  # Hello, Alice!
print(say_hi("Bob"))       # Hi, Bob.
```

## Higher-Order Functions

Functions that take or return functions.

```python
# map, filter, reduce are higher-order functions
from functools import reduce

numbers = [1, 2, 3, 4, 5]

# map: apply function to each element
doubled = list(map(lambda x: x * 2, numbers))
print(f"Doubled: {doubled}")

# filter: keep elements where function returns True
evens = list(filter(lambda x: x % 2 == 0, numbers))
print(f"Evens: {evens}")

# reduce: accumulate into single value
total = reduce(lambda acc, x: acc + x, numbers, 0)
print(f"Sum: {total}")
```

```python
# But list comprehensions are often cleaner:
doubled = [x * 2 for x in numbers]
evens = [x for x in numbers if x % 2 == 0]
total = sum(numbers)

print(f"Doubled: {doubled}")
print(f"Evens: {evens}")
print(f"Sum: {total}")
```

## AI Code Pattern: Callback Functions

```python
def process_items(items, on_item=None, on_complete=None):
    """Process items with optional callbacks."""
    results = []
    
    for item in items:
        result = item * 2  # Some processing
        results.append(result)
        
        if on_item:  # Call callback for each item
            on_item(item, result)
    
    if on_complete:  # Call callback when done
        on_complete(results)
    
    return results

# Use with callbacks
def log_item(item, result):
    print(f"Processed {item} -> {result}")

def summarize(results):
    print(f"Done! Total: {sum(results)}")

process_items([1, 2, 3], on_item=log_item, on_complete=summarize)
```

## Summary

| Concept | Description |
|---------|-------------|
| Closure | Function that remembers outer scope |
| `nonlocal` | Modify variable in outer scope |
| Lambda | Anonymous function `lambda x: expr` |
| `functools.partial` | Pre-fill function arguments |
| Higher-order function | Takes/returns functions |
| Callback | Function passed to be called later |
